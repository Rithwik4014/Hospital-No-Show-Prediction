```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/KaggleV2-May-2016.csv")

print(df.shape)
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='No-show')

plt.title('Hospital Appointment Attendance')
plt.xlabel('Appointment Status')
plt.ylabel('Number of Appointments')

plt.show()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nInvalid ages:")
print(df[df['Age'] < 0]['Age'].value_counts())

print("\nAge range:")
print(df['Age'].min(), "to", df['Age'].max())
```

```python
df = df[df['Age'] >= 0]

print("Updated dataset shape:", df.shape)
print("Minimum age:", df['Age'].min())
```

```python
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])

print(df[['ScheduledDay', 'AppointmentDay']].dtypes)
```

```python
df['WaitingDays'] = (
    df['AppointmentDay'].dt.date - df['ScheduledDay'].dt.date
).apply(lambda x: x.days)

print(df['WaitingDays'].describe())

print("Negative waiting days:", (df['WaitingDays'] < 0).sum())
```

```python
df = df[df['WaitingDays'] >= 0]

print("Updated dataset shape:", df.shape)
print("Minimum waiting days:", df['WaitingDays'].min())
```

```python
waiting_analysis = df.groupby('WaitingDays')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(waiting_analysis.head(15))
```

```python
plt.figure(figsize=(10, 5))

plt.plot(waiting_analysis.index, waiting_analysis.values)

plt.title('No-Show Rate vs Waiting Days')
plt.xlabel('Waiting Days')
plt.ylabel('No-Show Rate (%)')

plt.grid(True)
plt.show()
```

```python
waiting_counts = df['WaitingDays'].value_counts().sort_index()

print(waiting_counts.tail(20))
```

```python
bins = [-1, 0, 7, 30, 60, 90, 180]

labels = ['Same Day', '1-7 Days', '8-30 Days',
          '31-60 Days', '61-90 Days', 'Above 90 Days']

df['WaitingGroup'] = pd.cut(
    df['WaitingDays'],
    bins=bins,
    labels=labels
)

waiting_summary = df.groupby('WaitingGroup', observed=True).agg(
    Total_Appointments=('No-show', 'count'),
    No_Show_Rate=('No-show', lambda x: (x == 'Yes').mean() * 100)
)

print(waiting_summary)
```

```python
df['Scheduled_DayOfWeek'] = df['ScheduledDay'].dt.dayofweek
df['Appointment_DayOfWeek'] = df['AppointmentDay'].dt.dayofweek

df['Appointment_Month'] = df['AppointmentDay'].dt.month
df['Appointment_Day'] = df['AppointmentDay'].dt.day

df['Is_Weekend'] = df['Appointment_DayOfWeek'].isin([5, 6]).astype(int)

print(df[['Scheduled_DayOfWeek', 'Appointment_DayOfWeek',
          'Appointment_Month', 'Appointment_Day',
          'Is_Weekend']].head())

print("\nUpdated dataset shape:", df.shape)
```

```python
weekday_analysis = df.groupby('Appointment_DayOfWeek')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(weekday_analysis)

weekday_analysis.plot(kind='bar', figsize=(8, 4))

plt.title('No-Show Rate by Appointment Weekday')
plt.xlabel('Weekday (0 = Monday, 6 = Sunday)')
plt.ylabel('No-Show Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
age_bins = [0, 12, 18, 30, 45, 60, 75, 120]

age_labels = [
    '0-12', '13-18', '19-30', '31-45',
    '46-60', '61-75', '76+'
]

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

age_analysis = df.groupby('AgeGroup', observed=True)['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(age_analysis)

age_analysis.plot(kind='bar', figsize=(9, 4))
plt.title('No-Show Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('No-Show Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
sms_analysis = df.groupby('SMS_received')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(sms_analysis)

sms_analysis.plot(kind='bar', figsize=(6, 4))
plt.title('No-Show Rate by SMS Reminder')
plt.xlabel('SMS Received (0 = No, 1 = Yes)')
plt.ylabel('No-Show Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
medical_columns = ['Hipertension', 'Diabetes', 'Alcoholism']

for col in medical_columns:
    result = df.groupby(col)['No-show'].apply(
        lambda x: (x == 'Yes').mean() * 100
    )

    print(f"\nNo-Show Rate by {col}:")
    print(result)
```

```python
gender_analysis = df.groupby('Gender')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print("No-Show Rate by Gender:")
print(gender_analysis)

gender_analysis.plot(kind='bar', figsize=(6, 4))

plt.title('No-Show Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('No-Show Rate (%)')
plt.xticks(
    ticks=[0, 1],
    labels=['Female (F)', 'Male (M)'],
    rotation=0
)
plt.tight_layout()
plt.show()
```

```python
neighbourhood_translation = {
    'ILHAS OCEÂNICAS DE TRINDADE': 'Trindade Oceanic Islands',
    'SANTOS DUMONT': 'Santos Dumont',
    'SANTA CECÍLIA': 'Santa Cecilia',
    'SANTA CLARA': 'Santa Clara',
    'ITARARÉ': 'Itarare',
    'JESUS DE NAZARETH': 'Jesus of Nazareth',
    'HORTO': 'Garden',
    'ILHA DO PRÍNCIPE': 'Prince Island',
    'CARATOÍRA': 'Caratoira',
    'ANDORINHAS': 'Swallows',
    'PRAIA DO SUÁ': 'Sua Beach',
    'GURIGICA': 'Gurigica',
    'BENTO FERREIRA': 'Bento Ferreira',
    'PARQUE MOSCOSO': 'Moscoso Park',
    'MARUÍPE': 'Maruipe'
}

neighbourhood_analysis = df.groupby('Neighbourhood')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
).sort_values(ascending=False).head(15)

neighbourhood_analysis.index = [
    neighbourhood_translation.get(name, name)
    for name in neighbourhood_analysis.index
]

plt.figure(figsize=(12, 6))
neighbourhood_analysis.plot(kind='bar')

plt.title('Top 15 Neighbourhoods by No-Show Rate')
plt.xlabel('Neighbourhood')
plt.ylabel('No-Show Rate (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
neighbourhood_summary = df.groupby('Neighbourhood').agg(
    Total_Appointments=('No-show', 'count'),
    No_Shows=('No-show', lambda x: (x == 'Yes').sum())
)

neighbourhood_summary['No_Show_Rate'] = (
    neighbourhood_summary['No_Shows'] /
    neighbourhood_summary['Total_Appointments'] * 100
)

neighbourhood_summary = neighbourhood_summary[
    neighbourhood_summary['Total_Appointments'] >= 100
]

top_neighbourhoods = neighbourhood_summary.sort_values(
    by='No_Show_Rate',
    ascending=False
).head(15)

print(top_neighbourhoods.round(2))
```

```python
plt.figure(figsize=(12, 6))

top_neighbourhoods['No_Show_Rate'].sort_values().plot(
    kind='barh'
)

plt.title('Top 15 Neighbourhoods by No-Show Rate')
plt.xlabel('No-Show Rate (%)')
plt.ylabel('Neighbourhood')
plt.tight_layout()
plt.show()
```
